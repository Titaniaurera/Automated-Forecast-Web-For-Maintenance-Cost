```python
import pandas as pd


from preprocessing import preprocessing

from movement_cleansing import (
    clean_gi_movement
)

from readiness import readiness
```

```python
GI_FILE = r"D:\Forecasting SIG Project\ZCMM0167_GI.xlsx"


gi_raw = pd.read_excel(
    GI_FILE,
    sheet_name="Sheet1"
)


print("="*60)
print("GI RAW")
print("="*60)


print(
    "Shape:",
    gi_raw.shape
)


print(
    "Material Unik:",
    gi_raw["Material"].nunique()
)
```

```python
gi = preprocessing(
    gi_raw
)


print("="*60)
print("AFTER GI PREPROCESSING")
print("="*60)


print(
    "Jumlah Row:",
    len(gi)
)


print(
    "Jumlah Material:",
    gi["Material"].nunique()
)
```

```python
gi_clean = clean_gi_movement(
    gi
)


print("="*60)
print("AFTER GI MOVEMENT CLEANSING")
print("="*60)


print(
    "Jumlah Row:",
    len(gi_clean)
)


print(
    "Jumlah Material:",
    gi_clean["Material"].nunique()
)
```

```python
gi_ready = readiness(
    gi_clean
)


print("="*60)
print("AFTER GI READINESS")
print("="*60)


print(
    "Jumlah Row:",
    len(gi_ready)
)


print(
    "Jumlah Material:",
    gi_ready["Material"].nunique()
)



# =====================================================
# EXPORT SEMUA MATERIAL SETELAH READINESS
# =====================================================


gi_material_list = (

    gi_ready[

        [

            "Material",

            "Material Description"

        ]

    ]

    .drop_duplicates()

    .sort_values(
        by="Material"
    )

)



print(
    "Jumlah Material Export:",
    len(gi_material_list)
)



# Simpan CSV

gi_material_list.to_csv(

    "GI_AFTER_READINESS_MATERIAL.csv",

    index=False

)



print(
    "File berhasil dibuat:"
)

print(
    "GI_AFTER_READINESS_MATERIAL.csv"
)



print("="*60)
```

```python
GR_FILE = r"D:\Forecasting SIG Project\ZCMM0167_GR.xlsx"


gr_raw = pd.read_excel(
    GR_FILE,
    sheet_name="Sheet1"
)


print("="*60)
print("GR RAW")
print("="*60)


print(
    "Jumlah Row:",
    len(gr_raw)
)


print(
    "Jumlah Material:",
    gr_raw["Material"].nunique()
)
```

```python
gr = preprocessing(
    gr_raw
)


print("="*60)
print("AFTER GR PREPROCESSING")
print("="*60)


print(
    "Jumlah Row:",
    len(gr)
)


print(
    "Jumlah Material:",
    gr["Material"].nunique()
)
```

```python
gr_ready = readiness(
    gr
)


print("="*60)
print("AFTER GR READINESS")
print("="*60)


print(
    "Jumlah Row:",
    len(gr_ready)
)


print(
    "Jumlah Material:",
    gr_ready["Material"].nunique()
)
```

```python
gr_material_list = (

    gr_ready[

        [

            "Material",

            "Material Description"

        ]

    ]

    .drop_duplicates()

    .sort_values(
        by="Material"
    )

)



print(
    "Jumlah Material Export:",
    len(gr_material_list)
)



gr_material_list.to_csv(

    "GR_AFTER_READINESS_MATERIAL.csv",

    index=False

)



print(
    "File berhasil dibuat:"
)

print(
    "GR_AFTER_READINESS_MATERIAL.csv"
)
```

```python
gi_material = set(
    gi_ready["Material"]
)


gr_material = set(
    gr_ready["Material"]
)



match = gi_material & gr_material



print("="*60)
print("MATCH MATERIAL GI VS GR")
print("="*60)


print(
    "GI:",
    len(gi_material)
)


print(
    "GR:",
    len(gr_material)
)


print(
    "Material Sama:",
    len(match)
)


print("="*60)



pd.DataFrame(
    sorted(match),
    columns=["Material"]
).to_csv(
    "MATCH_GI_GR_AFTER_READINESS.csv",
    index=False
)
```
