```python
import pandas as pd


STOCK_FILE = r"D:\Forecasting SIG Project\3 JULI 2026.xlsx"



# =====================================================
# LOAD STOCK TERKINI
# =====================================================


stock_raw = pd.read_excel(

    STOCK_FILE,

    sheet_name="Sheet1"

)



print("="*60)
print("STOCK RAW")
print("="*60)



print(
    "Shape:",
    stock_raw.shape
)



print(
    "Jumlah Material:",
    stock_raw["Material"].nunique()
)






# =====================================================
# NORMALISASI MATERIAL STOCK
# =====================================================


stock_raw["Material"] = (

    stock_raw["Material"]

    .astype(str)

    .str.strip()

    .str.replace(
        ".0",
        "",
        regex=False
    )

)





stock_material = set(

    stock_raw["Material"]

    .unique()

)





print("="*60)
print("STOCK AFTER NORMALIZATION")
print("="*60)



print(
    "Jumlah Material Unik Stock:",
    len(stock_material)
)






# =====================================================
# MATCH GI + GR + STOCK
# =====================================================


match_stock = (

    match

    &

    stock_material

)





print("="*60)
print("MATCH GI + GR + STOCK")
print("="*60)



print(
    "GI Material:",
    len(gi_material)
)



print(
    "GR Material:",
    len(gr_material)
)



print(
    "GI + GR:",
    len(match)
)



print(
    "GI + GR + STOCK:",
    len(match_stock)
)



print("="*60)



print(
    "Contoh Material Match:"
)



print(
    list(match_stock)[:50]
)
```

```python
# =====================================================
# FILTER MATERIAL MATCH GI + GR + STOCK
# DENGAN HISTORI MINIMAL 2
# =====================================================


# material yang sudah match 4238
match_material = set(
    match_stock
)



# =====================================================
# HISTORI GI
# =====================================================

gi_hist = (

    gi_final

    [
        gi_final["Material"]
        .isin(match_material)
    ]

    .groupby("Material")

    .size()

    .reset_index(
        name="GI_History"
    )

)



# =====================================================
# HISTORI GR
# =====================================================

gr_hist = (

    gr_final

    [
        gr_final["Material"]
        .isin(match_material)
    ]

    .groupby("Material")

    .size()

    .reset_index(
        name="GR_History"
    )

)





# =====================================================
# GABUNG GI + GR
# =====================================================

history_match = gi_hist.merge(

    gr_hist,

    on="Material",

    how="inner"

)




# =====================================================
# FILTER MINIMAL HISTORI 2
# =====================================================

history_valid = history_match[

    (history_match["GI_History"] >= 2)

    &

    (history_match["GR_History"] >= 2)

]





print("="*60)

print("FILTER MATCH GI + GR + STOCK")

print("="*60)



print(
    "Awal GI+GR+STOCK:",
    len(match_material)
)



print(
    "Setelah GI >= 2 dan GR >= 2:",
    len(history_valid)
)



print("="*60)



print(
    history_valid.head(20)
)
```
